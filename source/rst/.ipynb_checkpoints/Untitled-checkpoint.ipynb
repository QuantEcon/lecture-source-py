```python
import glob
import errno

path = '*.rst'
#path = "*.rst"
files = glob.glob(path)
```

```python
def make_texList(filename):
    f = open(filename, "r", encoding="utf8")
    texList = []
    for line in f:
        texList.append(line)
    return texList
```

```python
def func_replace(a, b, filename):
    with open(filename,'r+', encoding="utf8") as f:
            #convert to string:
            data = f.read()
            f.seek(0)
            f.write(data.replace(a, b))
            f.truncate()
```
